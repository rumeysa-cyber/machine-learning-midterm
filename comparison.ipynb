# 11. Comparing the Two Datasets

```python

import pandas as pd

# Load both classifier summary files
breast_summary = pd.read_csv(
    "results/breast_cancer_classifier_summary.csv"
)

wine_summary = pd.read_csv(
    "results/wine_classifier_summary.csv"
)

# Combine both datasets
combined_summary = pd.concat(
    [breast_summary, wine_summary],
    ignore_index=True
)

print("TOTAL CONFIGURATIONS:", len(combined_summary))

assert len(breast_summary) == 22
assert len(wine_summary) == 22
assert len(combined_summary) == 44

print("\nBREAST CANCER - TOP 5 CONFIGURATIONS")
display(breast_summary.head(5))

print("\nWINE - TOP 5 CONFIGURATIONS")
display(wine_summary.head(5))

print("\nBOTH DATASETS COMBINED")
display(combined_summary)
```

## 11.1 Classification and Scaling Comparison
## 11.2 Parameter Effects and Seed Variability
## 11.3 Classification Errors and K-Means Comparison

```python

import pandas as pd
import numpy as np

# Load final classifier test results
breast_test = pd.read_csv(
    "results/breast_cancer_classifier_test.csv"
)

wine_test = pd.read_csv(
    "results/wine_classifier_test.csv"
)

# Load final K-Means test results
breast_kmeans = pd.read_csv(
    "results/breast_cancer_kmeans_test.csv"
)

wine_kmeans = pd.read_csv(
    "results/wine_kmeans_test.csv"
)

# 11.1 Classification Comparison
classifier_test_comparison = pd.concat(
    [breast_test, wine_test],
    ignore_index=True
)

print("1. FINAL CLASSIFICATION COMPARISON")

display(
    classifier_test_comparison[
        [
            "dataset",
            "selected_configuration",
            "mean_val_macro_f1",
            "std_val_macro_f1",
            "test_accuracy",
            "test_macro_f1",
            "test_roc_auc"
        ]
    ].round(4)
)

# Scaling effects using validation macro-F1
print("\n2. SCALING EFFECTS")

scaling_comparison = combined_summary.pivot_table(
    index=["dataset", "algorithm", "parameter"],
    columns="representation",
    values="val_macro_f1_mean"
).reset_index()

scaling_comparison["scaled_minus_unscaled"] = (
    scaling_comparison["scaled"] -
    scaling_comparison["unscaled"]
)

display(scaling_comparison.round(4))

print("\nAVERAGE SCALING EFFECT BY ALGORITHM")

scaling_effect_summary = (
    scaling_comparison.groupby(
        ["dataset", "algorithm"],
        as_index=False
    )["scaled_minus_unscaled"]
    .mean()
)

display(
    scaling_effect_summary.round(4)
)

# 11.2 Parameter effects
print("\n3. DECISION TREE DEPTH COMPARISON")

tree_comparison = combined_summary[
    combined_summary["algorithm"] == "decision_tree"
].copy()

display(
    tree_comparison[
        [
            "dataset",
            "representation",
            "parameter",
            "train_macro_f1_mean",
            "val_macro_f1_mean",
            "val_macro_f1_std"
        ]
    ].round(4)
)

print("\n4. KNN NEIGHBOUR COMPARISON")

knn_comparison = combined_summary[
    combined_summary["algorithm"] == "knn"
].copy()

display(
    knn_comparison[
        [
            "dataset",
            "representation",
            "parameter",
            "train_macro_f1_mean",
            "val_macro_f1_mean",
            "val_macro_f1_std"
        ]
    ].round(4)
)

# Seed variability for all configurations
print("\n5. VALIDATION VARIABILITY BY DATASET")

variability_summary = (
    combined_summary.groupby("dataset")
    ["val_macro_f1_std"]
    .agg(["mean", "min", "max"])
)

display(variability_summary.round(4))

# 11.3 K-Means comparison
print("\n6. FINAL K-MEANS COMPARISON")

kmeans_test_comparison = pd.concat(
    [breast_kmeans, wine_kmeans],
    ignore_index=True
)

display(
    kmeans_test_comparison[
        [
            "dataset",
            "representation",
            "evaluation_seed",
            "test_ari",
            "test_silhouette"
        ]
    ].round(4)
)

```

# 12. Required Run and Output Inventory
## 12.1 Verify Model Run Counts
## 12.2 Export Six Final CSV Files

```python

import pandas as pd
from pathlib import Path

# Final output directory
results_dir = Path("results")
results_dir.mkdir(exist_ok=True)

# Required filenames and expected row counts
required_files = {
    "classifier_runs": 132,
    "classifier_summary": 44,
    "classifier_test": 2,
    "kmeans_runs": 12,
    "kmeans_summary": 4,
    "kmeans_test": 4
}

print("12.1 VERIFYING AND COMBINING EXPERIMENT RESULTS")

for file_type, expected_rows in required_files.items():

    # Read each dataset's saved results
    breast_file = results_dir / f"breast_cancer_{file_type}.csv"
    wine_file = results_dir / f"wine_{file_type}.csv"

    breast_df = pd.read_csv(breast_file)
    wine_df = pd.read_csv(wine_file)

    # Combine both datasets
    combined_df = pd.concat(
        [breast_df, wine_df],
        ignore_index=True
    )

    # Verify required dataset labels
    assert "dataset" in combined_df.columns
    assert set(combined_df["dataset"]) == {
        "breast_cancer", "wine"
    }

    # Verify row count
    actual_rows = len(combined_df)

    assert actual_rows == expected_rows, (
        f"{file_type}: Expected {expected_rows} rows, "
        f"found {actual_rows}"
    )

    # Verify no completely duplicated rows
    assert not combined_df.duplicated().any(), (
        f"Duplicate rows detected in {file_type}"
    )

    # Export final combined CSV
    output_path = results_dir / f"{file_type}.csv"

    combined_df.to_csv(
        output_path,
        index=False
    )

    print(
        f"{file_type}.csv: "
        f"{actual_rows}/{expected_rows} rows - OK"
    )

print("\n12.2 CHECKING TOTAL MODEL FITS")

classifier_runs = pd.read_csv(
    results_dir / "classifier_runs.csv"
)

kmeans_runs = pd.read_csv(
    results_dir / "kmeans_runs.csv"
)

assert classifier_runs["experiment_id"].is_unique
assert kmeans_runs["experiment_id"].is_unique

total_fits = len(classifier_runs) + len(kmeans_runs)

assert total_fits == 144

print("Classifier fits:", len(classifier_runs))
print("K-Means fits:", len(kmeans_runs))
print("TOTAL MODEL FITS:", total_fits)

```

```python

import pandas as pd

df = pd.read_csv("results/classifier_summary.csv")

# Average scaling effect for each algorithm
scaling = df.pivot_table(
    index=["dataset", "algorithm", "parameter"],
    columns="representation",
    values="val_macro_f1_mean"
).reset_index()

scaling["difference"] = scaling["scaled"] - scaling["unscaled"]

print("1. SCALING EFFECTS")
display(
    scaling.groupby(["dataset", "algorithm"])["difference"]
    .mean().round(4)
)

# Tree depth and KNN parameter results
for dataset in ["breast_cancer", "wine"]:
    print(f"\n2. PARAMETER RESULTS - {dataset.upper()}")

    for algorithm in ["decision_tree", "knn"]:
        print(f"\n{algorithm.upper()}")

        subset = df[
            (df["dataset"] == dataset) &
            (df["algorithm"] == algorithm)
        ]

        display(
            subset[
                ["representation", "parameter",
                 "val_macro_f1_mean", "val_macro_f1_std"]
            ].round(4)
        )

print("\n3. SEED VARIABILITY")
display(
    df.groupby("dataset")["val_macro_f1_std"]
    .agg(["mean", "min", "max"])
    .round(4)
)
```
